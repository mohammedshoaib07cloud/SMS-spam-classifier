```python
df=df.drop_duplicates()
df.shape
```

```python
df["length"]=df["message"].apply(len)
df.groupby("label")["length"].describe()
```

```python
df["label"].value_counts(normalize=True)
df.shape
```

```python
df["label_num"]=df["label"].map({"ham":0,"spam":1}).copy()
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df["message"],
    df["label_num"],
    test_size=0.2,
    random_state=42,
    stratify=df["label_num"]
)

print(X_train.shape, X_test.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words="english", lowercase=True)

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print(X_train_vec.shape)
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_vec, y_train)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

y_pred = model.predict(X_test_vec)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print()
print(classification_report(y_test, y_pred, target_names=["ham", "spam"]))
```

```python
results = pd.DataFrame({
    "message": X_test,
    "actual":y_test,
    "predicted":y_pred
})

results[results["actual"] != results["predicted"]]
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(class_weight="balanced", max_iter=1000)
lr.fit(X_train_vec, y_train)
y_pred_lr = lr.predict(X_test_vec)

print(classification_report(y_test, y_pred_lr, target_names=["ham", "spam"]))
```

```python
import numpy as np
from scipy.sparse import hstack

def extra_features(texts):
    return np.array([
        [
            sum(c.isdigit() for c in t),                 # digit count
            sum(c.isupper() for c in t) / max(len(t), 1), # share of capitals
            len(t),                                       # length
        ]
        for t in texts
    ])

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

train_extra = scaler.fit_transform(extra_features(X_train))
test_extra = scaler.transform(extra_features(X_test))

X_train_combo = hstack([X_train_vec, train_extra])
X_test_combo = hstack([X_test_vec, test_extra])

lr2 = LogisticRegression(class_weight="balanced", max_iter=1000)
lr2.fit(X_train_combo, y_train)
y_pred_lr2 = lr2.predict(X_test_combo)

print(classification_report(y_test, y_pred_lr2, target_names=["ham", "spam"]))
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models = {
    "TF-IDF + Naive Bayes": y_pred,
    "TF-IDF + Logistic Regression": y_pred_lr,
    "+ extra features": y_pred_lr2,
}

rows = []
for name, preds in models.items():
    rows.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, preds),
        "Precision": precision_score(y_test, preds),
        "Recall": recall_score(y_test, preds),
        "F1": f1_score(y_test, preds),
    })

comparison = pd.DataFrame(rows).round(3)
comparison
```
